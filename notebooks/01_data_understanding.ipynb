## Conclusions

### Dataset Overview

- The dataset contains 200,002 flight observations and 20 variables.
- It covers flights departing between May 1, 2026 and June 30, 2026.
- The dataset includes information about airlines, routes, airports, departure and arrival times, number of stops, flight duration, booking lead time, and ticket prices.
- The target variable for the predictive modeling stage will be 'price'.

### Data Quality

- No missing values were detected across any of the 20 variables.
- The duplicate rate is approximately 0.0005%, indicating that duplicated observations are negligible.
- Several temporal variables (departure_time , arrival_time, and scraped_at) are currently stored as strings and should be converted to datetime format.
- 'duration' is also stored as a string and will require inspection and potentially conversion into a numerical representation.

### Dataset Characteristics

- The dataset contains flights from 41 source countries and 40 destination countries, providing broad geographical coverage across Europe.
- 'days_left' contains 61 distinct values and represents the booking lead time before departure.
- Departure and arrival dayparts contain four categories, while day-of-week variables contain the expected seven categories.

### Target Variable

- Ticket prices range from 17 to 11,614.
- The median ticket price is 289, while the mean is approximately 384.
