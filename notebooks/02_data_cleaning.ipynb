# Airfare Analytics — Data Cleaning

This notebook focuses on cleaning and validating the raw airfare dataset before exploratory data analysis.


```python
# Import necessary libraries
from pathlib import Path

import pandas as pd
import numpy as np
```

```python
# Load the dataset
DATA_PATH = Path("../data/raw/europe_flights_google_prices.csv")

airfare_df = pd.read_csv(DATA_PATH)
```

```python
#Check the shape of the dataset
airfare_df.shape
```

```python
airfare_df.head()
```

```python
# Bring the information about the dataset
airfare_df.dtypes
```

```python
# Convert the 'date' column to datetime format
airfare_df['arrival_time'] = pd.to_datetime(airfare_df['arrival_time'])
airfare_df['departure_time'] = pd.to_datetime(airfare_df['departure_time'])
airfare_df['scraped_at'] = pd.to_datetime(airfare_df['scraped_at'])
```

```python
#Check the data types again after conversion
airfare_df.dtypes
```

```python
airfare_df.head()
```

```python
# Check for duplicates in the dataset
airfare_df[airfare_df.duplicated(keep=False)]
```

```python
#Drop duplicates from the dataset
airfare_df = airfare_df.drop_duplicates()
```

```python
#Check for duplicates in the dataset
print('Number of duplicates:', airfare_df.duplicated().sum())
```

```python
#Check for missing values in the dataset
#airfare_df.isnull().sum()
```

```python
# Convert the 'duration' column to total minutes
duration_td = pd.to_timedelta(airfare_df["duration"])
airfare_df["duration_minutes"] = duration_td.dt.total_seconds() / 60
airfare_df[["duration", "duration_minutes"]].head(10)
```

```python
# After the change in the duration column, we can drop the original 'duration' column as it is no longer needed.
airfare_df = airfare_df.drop(columns=["duration"])
```

```python
# As a final step, we will save the cleaned dataset to a new CSV file for further analysis.
OUTPUT_PATH = Path("../data/processed/airfare_clean.csv")

airfare_df.to_csv(OUTPUT_PATH, index=False)

print(f"Clean dataset saved to: {OUTPUT_PATH}")
print(f"Final shape: {airfare_df.shape}")
```

## Conclusions

The Data Cleaning stage confirmed that the dataset was already in good overall condition and required only limited preprocessing.

### Cleaning performed

- Temporal variables were converted to appropriate datetime formats.
- Flight duration was converted to a numerical representation suitable for analysis.
- Exact duplicated observations were identified and removed.
- Missing values were checked, with no missing observations detected.
- Variable types and basic value consistency were validated.

### Final dataset

The cleaned dataset was saved as `data/processed/airfare_clean.csv`.
